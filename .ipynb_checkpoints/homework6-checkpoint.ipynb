```python
def isprime(x):
    primerows = []
    for row in x:
        for n in row:
            if n < 2:
                continue 
            prime = True
            for i in range(2, n):
                if n % i == 0:
                    prime = False
                    break
            if prime:
                primerows.append(row)
                break
    return(np.array(primerows))
   
```

```python
def zeroarray():
    return np.zeros((8,8))
```

```python
def checkerboardodd():
    arr = np.zeros((8,8))
    arr[::2, ::2] = 1
    return arr
```

```python
def checkerboard():
    arr = np.zeros((8,8))
    arr[::2, ::2] = 1
    arr[1::2, 1::2] = 1
    return arr
```

```python
def reversecheckerboard():
    arr = np.zeros((8,8))
    arr[1::2, ::2] = 1
    arr[::2, 1::2] = 1
    return arr
```

```python
def expansion(arr, n):
    spaces = " " * n
    result = []
    for x in arr:
        result.append(spaces.join(x))
    return np.array(result)
universe = np.array(['galaxy', 'clusters'])
```

```python
def secondbrightest(arr):
    result = []
    for row in arr:
        sortedrow = np.sort(row)
        result.append(sortedrow[-2])
    return np.array(result)
        
```
