```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = np.loadtxt("../../course_assignments/homework7/star_data.txt", skiprows=1)
```

```python
print(data.shape)
```

```python
magnitude = data[:,0]
```

```python
temperature = data[:,1]
```

```python
print(magnitude.max())
#corresponds to the star with the largest magnitude 
```

```python
print(magnitude.min())
#corresponds to the star with the smallest magnitude
```

```python
print(temperature.max())
#corresponds to the hottest star
```

```python
print(temperature.min())
#corresponds to the coldest star
```

```python
#the brightest star is the one with magnitude 2.24 and dimmest is 12.49
```

```python
log_magnitude = np.log(magnitude)
log_temperature = np.log(temperature)
```

```python
plt.figure(figsize = (8,6))
plt.scatter(log_temperature, log_magnitude)
```

```python
plt.figure(figsize=(8,6))
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
```

```python
b = np.argmin(magnitude)
d = np.argmax(magnitude)
h = np.argmax(temperature)
c = np.argmin(temperature)
plt.figure(figsize = (8,6))
plt.scatter(log_temperature, log_magnitude)
plt.gca().invert_xaxis()
plt.gca().invert_yaxis()
plt.scatter(log_temperature[b], log_magnitude[b], color = "black", marker="*", label = "brightest", s = 300)
plt.scatter(log_temperature[d], log_magnitude[d], color ="purple", marker="*", label = "dimmest", s = 300)
plt.scatter(log_temperature[h], log_magnitude[h], color ="green", marker="*", label = "hottest", s = 300)
plt.scatter(log_temperature[c], log_magnitude[c], color = "orange", marker="*", label = "coldest", s = 300)
plt.legend()
plt.title("Hertzsprung-Russel Diagram")
plt.xlabel("Temperature (log K)")
plt.ylabel("Magnitude")
```
