```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
def make_sine_wave(x, A, w):
    y = A*np.sin(w*x)
    return y
```

```python
x = np.linspace(0, 2 * np.pi, 1000)
```

```python
amplitude = np.array([0.5, 1, 1.5, 2, 2.5])
frequency = np.array([1, 2, 3, 4, 5])
```

```python
fig, axes = plt.subplots(1,3, figsize = (25,5))
for A, w in zip(amplitude, frequency):
    axes[0].plot(x, make_sine_wave(x, A, 1), label = f"A={A}")
    axes[1].plot(x, make_sine_wave(x, 1, w), label = f"w={w}")
    axes[2].plot(x, make_sine_wave(x, A, w), label = f"A={A}, w={w}")
axes[0].set_title("Amplitude")
axes[1].set_title("Frequency")
axes[2].set_title("Amp and Freq")
for ax in axes:
    ax.legend()
plt.show()
```
